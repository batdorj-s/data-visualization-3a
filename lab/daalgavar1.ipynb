## 1.1 Дүрслэлийн зарчмуудыг тодорхойлох

Доорх 3 зургийг тус тусад нь шинжилж, **асуудал → зарчмын зөрчил → засвар** тохиолдлыг гаргав.

### Зураг А: Bar chart (y тэнхлэг 95–100 хүртэл)

```python
subjects = ['Математик', 'Физик', 'Хими', 'Биологи', 'Мэдээлэл зүй']
scores = [97.4, 69.1, 80.2, 95.5, 98.9]

plt.figure(figsize=(8, 4.5))
plt.bar(subjects, scores, color='#7CAED6', width=0.6)
plt.ylim(95, 100)
plt.title("Зураг А: Шалгалтын дундаж оноо")
plt.ylabel("Оноо")
plt.grid(axis='y', linestyle='--', alpha=0.4)
plt.tight_layout()
plt.show()
```

**Асуудал:**
Y тэнхлэг **0-ээс биш, 95-аас** эхэлсэн. 95.5–98.9 хоорондын жижиг зөрүүгүүд асар том ялгаа мэт харагдана — үзэгч бодит зөрүүнээс хэд дахин их мэдэрнэ.

**Зарчмын зөрчил:**
- **Пропорциональ хэмжээст байдал** (Tufte: "lie factor" = графикаас мэдрэх хэмжээ / өгөгдлийн бодит хэмжээ). Bar chart-ийн суурь **0 байх ёстой**; 0-ээс тасарсан тэнхлэг харьцуулалтыг гажуудуулна.
- **Шударга харьцуулалт** зөрчлөө: баганын өндөр нь жинхэнэ утгын хэмжээтэй пропорциональ биш.

**Засвар:**
- Y тэнхлэг **0-ээс эхлүүлэх** (`plt.ylim(0, 100)`).
- Жижиг ялгааг харуулах шаардлагатай бол баганан графикаас өөр хэрэгсэл сонгох: **dot plot + утгын label**, эсвэл тасархай (broken) тэнхлэгийг тодорхүy тэмдэглэх.

```python
import matplotlib.pyplot as plt

# Өгөгдөл
subjects = ['Математик', 'Физик', 'Хими', 'Биологи', 'Мэдээлэл зүй']
scores = [97.4, 69.1, 80.2, 95.5, 98.9]

# Зураг үүсгэх
plt.figure(figsize=(8, 5))
bars = plt.bar(subjects, scores, color='#7CAED6', width=0.6)

# Y тэнхлэгийг 0-100 хүртэл болгож зассан (Баганан диаграмм заавал 0-ээс эхлэх ёстой)
plt.ylim(0, 105) 

plt.title("Зураг А: Шалгалтын дундаж оноо", pad=15)
plt.ylabel("Оноо")
plt.grid(axis='y', linestyle='--', alpha=0.4)

# Багана бүрийн орой дээр нарийвчилсан утгыг нь бичиж харуулах
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval + 1, round(yval, 1), 
             ha='center', va='bottom', fontsize=10, fontweight='bold')

# X тэнхлэгийн бичвэрүүд урт байвал давхарцахгүй байх үүднээс бага зэрэг хазгай болгох
plt.xticks(rotation=15)

# Зургийн хүрээг цэгцтэй болгох
plt.tight_layout()
plt.show()
```

### Зураг Б: Pie chart (15 сектор бүхий)

```python
labels = [f'Сектор {i}' for i in range(1, 16)]
values = [8, 12, 5, 15, 7, 10, 6, 9, 11, 4, 13, 8, 7, 12, 10]

plt.figure(figsize=(7, 5.5))
plt.pie(values, labels=labels, autopct='%1.1f%%', startangle=140,
        colors=plt.cm.tab20.colors[:15],
        wedgeprops={'edgecolor': 'white', 'linewidth': 1})
plt.title("Зураг Б: 15 сектортой Pie chart")
plt.axis('equal')
plt.tight_layout()
plt.show()
```

**Асуудал:**
15 сектор маш олон. Хүний нүд **өнцөг, талбайг** нарийвчлан харьцуулж чаддаггүй тул 4% эсвэл 13% гэх мэт жижиг секторуудыг ялгахад хэцүү. График "хулгана-зөөмөл" (lebel бөөгнөрсөн) харагдана.

**Зарчмын зөрчил:**
- **Мэдрэхүйн чадварын зарчим** (Cleveland): хүмүүс **шулуун шугамын урт**-ыг хамгийн сайн, **өнцөг/талбай**-г хамгийн муу харьцуулдаг.
- **Data-ink / Энгийн байдал** (Tufte): 15 сектор pie chart-д тохиромжгүй — pie нь **5–6-аас бага** хэсэгт л үр дүнтэй.

**Засвар:**
- **Эрэмбэлсэн хэвтээ bar chart** болгох (секторыг шулуун уртаар харьцуулдаг).
- Эсвэл жижиг секторуудыг нэгтгэж **"Бусад"** категори үүсгээд хэдхэн (5 орчим) гол хэсэг үлдээх.

```python
import matplotlib.pyplot as plt
import numpy as np

# Өгөгдөл
labels = [f'Сектор {i}' for i in range(1, 16)]
values = [8, 12, 5, 15, 7, 10, 6, 9, 11, 4, 13, 8, 7, 12, 10]

# ==========================================
# ХУВИЛБАР 1: Эрэмбэлсэн хэвтээ Bar Chart
# ==========================================
# Өгөгдлийг өсөхөөр эрэмбэлэх
sorted_indices = np.argsort(values)
sorted_labels = [labels[i] for i in sorted_indices]
sorted_values = [values[i] for i in sorted_indices]

plt.figure(figsize=(9, 6))
bars = plt.barh(sorted_labels, sorted_values, color='#7CAED6')
plt.title("Хувилбар 1: Эрэмбэлсэн хэвтээ Bar Chart", pad=15)
plt.xlabel("Утга")
plt.grid(axis='x', linestyle='--', alpha=0.5)

# Утгуудыг баганын төгсгөлд харуулах
for bar in bars:
    width = bar.get_width()
    plt.text(width + 0.2, bar.get_y() + bar.get_height()/2, 
             str(width), ha='left', va='center', fontsize=9)

plt.tight_layout()
plt.show()

# ==========================================
# ХУВИЛБАР 2: "Бусад" категори үүсгэсэн Pie Chart
# ==========================================
# Өгөгдлийг буурахаар эрэмбэлэх
sorted_desc_indices = np.argsort(values)[::-1]
top_n = 5 # Зөвхөн топ 5-ыг үлдээх

top_indices = sorted_desc_indices[:top_n]
other_indices = sorted_desc_indices[top_n:]

# Шинэ шошго болон утгуудыг үүсгэх
new_labels = [labels[i] for i in top_indices] + ["Бусад"]
new_values = [values[i] for i in top_indices] + [sum(values[i] for i in other_indices)]

plt.figure(figsize=(7, 5.5))
plt.pie(new_values, labels=new_labels, autopct='%1.1f%%', startangle=140,
        colors=plt.cm.Pastel1.colors[:len(new_labels)],
        wedgeprops={'edgecolor': 'white', 'linewidth': 1.5})
plt.title("Хувилбар 2: Топ 5 болон 'Бусад' ангилалтай Pie chart", pad=15)
plt.axis('equal')
plt.tight_layout()
plt.show()
```

### Зураг В: Line chart (5 өөр өнгийн шугам)

**Асуудал:**
5 шугам **зөвхөн өнгөөр** ялгагдана. Өнгө сохорт (color vision deficiency) хүмүүсийн ~8% — улаан-ногоон ялгах чадваргүй хэрэглэгч ялгах боломжгүй. Давхцсан шугамуудыг харьцуулах нь төвөгтэй.

**Зарчмын зөрчил:**
- **Илүүдэл (Redundant) кодчилол** дутагдсан: хэрэглэгч хуваагдах сувгаа зөвхөн **hue** дээр найруулсан.
- **Хүртээмж (Accessibility)**: өнгө нь цорын ганц ялгаа болвол хэсэг хэрэглэгч мэдээллийг алддаг.

**Засвар:**
- **Маркер + шугамын хэвлэл** (linestyle) нэмэх: `-`, `--`, `-.`, `:`, `o` ба `s`.
- Red-green биш **colorblind-safe** palette (ColorBrewer `Set2`, `Paired`) сонгох.
- Илүү олон шугам бол **жижиг граф (small multiples)** / facet ангилна.

## 1.2 Хүний мэдрэхүйн онцлогийг ашиглах

**Сценари:** 4 төрлийн бараа (A, B, C, D) × 3 улирал (Өвөл, Хавар, Зун) борлуулалтын өгөгдөл.

| Харааны суваг | Юуг кодлох | Яагаад |
|---|---|---|
| **Өнгө (hue)** | Улирал | Хүйтэн↔дулаан сэтгэлзүйн холбоо (pre-attentive) |
| **Хэлбэр (marker)** | Бараа төрөл | Өнгө сохорт хэрэглэгч ч ялгана (redundant encoding) |
| **Хэмжээ (size)** | Борлуулалтын дүн | Их = том, бага = жижиг |
| **Ирмэг (edge)** | Тодруулга | Давхцсан цэгүүдийг ялгана |

```python
import matplotlib.pyplot as plt
import numpy as np

# Өгөгдөл үүсгэх (өмнөх зассан хувилбартай ижил өгөгдөл)
np.random.seed(42)
years = np.arange(2015, 2025)
categories = ['Ангилал А', 'Ангилал Б', 'Ангилал В', 'Ангилал Г', 'Ангилал Д']
data = {}
for cat in categories:
    data[cat] = np.cumsum(np.random.randn(10) * 5) + np.random.randint(20, 80)

# Зураг үүсгэх
plt.figure(figsize=(9, 5))

# Бүх шугамыг өөр өнгөөр нэг график дээр давхарлаж зурах
for cat in categories:
    plt.plot(years, data[cat], marker='o', linewidth=2, label=cat)

plt.title("Зураг В: Алдаатай хувилбар (Spaghetti chart)", pad=15)
plt.xlabel("Он")
plt.ylabel("Утга")
plt.grid(True, linestyle='--', alpha=0.5)
plt.xticks(years)

# Тайлбарыг (Legend) тусад нь байрлуулах 
# (Энэ нь уншигч графикаас тайлбар руу байнга нүд шилжүүлэх шаардлага үүсгэдэг)
plt.legend() 

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Өгөгдөл үүсгэх
np.random.seed(42)
years = np.arange(2015, 2025)
categories = ['Ангилал А', 'Ангилал Б', 'Ангилал В', 'Ангилал Г', 'Ангилал Д']
data = {}
for cat in categories:
    data[cat] = np.cumsum(np.random.randn(10) * 5) + np.random.randint(20, 80)

# ==============================================================
# ХУВИЛБАР 1: Гол мэдээллийг онцлох болон шууд шошголох
# ==============================================================
fig, ax = plt.subplots(figsize=(10, 6))

# Чухал биш шугамуудыг саарал өнгөөр зурах
for cat in categories[2:]:
    ax.plot(years, data[cat], color='lightgray', linewidth=2, alpha=0.8)
    ax.text(years[-1] + 0.2, data[cat][-1], cat, color='gray', va='center', fontsize=10)

# Онцлох шугамуудыг тод өнгөөр зурах
ax.plot(years, data[categories[0]], color='#1f77b4', linewidth=3.5)
ax.text(years[-1] + 0.2, data[categories[0]][-1], categories[0], color='#1f77b4', va='center', fontweight='bold', fontsize=12)

ax.plot(years, data[categories[1]], color='#d62728', linewidth=3.5)
ax.text(years[-1] + 0.2, data[categories[1]][-1], categories[1], color='#d62728', va='center', fontweight='bold', fontsize=12)

ax.set_title('Хувилбар 1: Чухал мэдээллийг онцлох болон шууд шошголох', pad=15, fontsize=14)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.set_xticks(years)
ax.grid(axis='y', linestyle='--', alpha=0.3)
plt.xlim(years[0], years[-1] + 2.5) # Текст багтах зай гаргах
plt.tight_layout()
plt.show()

# ==============================================================
# ХУВИЛБАР 2: Жижиг олон график (Small multiples)
# ==============================================================
fig, axes = plt.subplots(5, 1, figsize=(8, 10), sharex=True, sharey=True)
fig.suptitle('Хувилбар 2: Жижиг олон график (Small multiples)', y=1.02, fontsize=14)

colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd']

for i, (cat, ax) in enumerate(zip(categories, axes)):
    # Бусад шугамыг бүдэг саарал дэвсгэр болгож харуулах (харьцуулах зорилгоор)
    for c in categories:
        ax.plot(years, data[c], color='lightgray', linewidth=1, alpha=0.5)
    
    # Тухайн графикийн гол шугамыг тод зурах
    ax.plot(years, data[cat], color=colors[i], linewidth=3)
    
    # Y тэнхлэгийн нэрийг хэвтээгээр байрлуулах
    ax.set_ylabel(cat, rotation=0, labelpad=40, ha='right', va='center', color=colors[i], fontweight='bold', fontsize=11)
    
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.grid(axis='y', linestyle='--', alpha=0.3)
    
    if i < 4:
        ax.spines['bottom'].set_visible(False)
        ax.tick_params(axis='x', length=0)
        
ax.set_xticks(years)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Өгөгдөл бэлтгэх
seasons = ['Өвөл', 'Хавар', 'Зун']
products = ['Бараа A', 'Бараа B', 'Бараа C', 'Бараа D']
colors = {'Өвөл': '#1f77b4', 'Хавар': '#2ca02c', 'Зун': '#ff7f0e'} 
shapes = {'Бараа A': 'o', 'Бараа B': 's', 'Бараа C': '^', 'Бараа D': 'D'} 

# Зохиомол борлуулалтын утгууд (Тест зорилгоор)
sales_data = {
    'Өвөл': {'Бараа A': 45, 'Бараа B': 80, 'Бараа C': 30, 'Бараа D': 60},
    'Хавар': {'Бараа A': 55, 'Бараа B': 40, 'Бараа C': 90, 'Бараа D': 70},
    'Зун': {'Бараа A': 85, 'Бараа B': 30, 'Бараа C': 50, 'Бараа D': 95}
}

plt.figure(figsize=(9, 6))

# График зурах
for season in seasons:
    for product in products:
        sales = sales_data[season][product]
        plt.scatter(season, sales, 
                    c=colors[season], 
                    marker=shapes[product], 
                    s=sales*15, # Борлуулалтын хэмжээгээр дүрсийн хэмжээг томруулах
                    alpha=0.7, 
                    edgecolors='black',
                    label=product if season == 'Өвөл' else "") 

# Legend тохиргоо 
handles, labels = plt.gca().get_legend_handles_labels()
by_label = dict(zip(labels, handles))

import copy
new_handles = []
for h in by_label.values():
    h_copy = copy.copy(h)
    h_copy.set_facecolor('gray')
    h_copy.set_sizes([150]) 
    new_handles.append(h_copy)

plt.legend(new_handles, by_label.keys(), title="Барааны төрөл (Хэлбэр)", loc='upper right')

plt.title('Хүний мэдрэхүйн онцлогийг ашигласан борлуулалтын график', fontsize=14, pad=15)
plt.xlabel('Улирал (Өнгөөр ялгасан)', fontsize=12)
plt.ylabel('Борлуулалтын хэмжээ (Y тэнхлэг ба Дүрсийн хэмжээ)', fontsize=12)
plt.ylim(0, 130)
plt.grid(True, linestyle='--', alpha=0.5)

plt.show()
```
