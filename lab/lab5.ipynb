## 1. Өгөгдлийн бүтэц ба хувьсагчид

Tominski, Schumann нарын тодорхойлолтоор **олон хувьсагчтай өгөгдөл** гэдэг нь нэг өгөгдлийн нэгж (энд — нэг зорчигч) дээр гурав ба түүнээс дээш хувьсагч $A = \{a_1, a_2, \dots, a_m\}$ хэмжигдсэн бүтэц юм. Хувьсагчийн **хэмжүүрийн түвшин** нь түүнийг ямар визуал сувгаар кодлох боломжтойг тодорхойлно:

| Түвшин | Хувьсагч | Утга | Тохирох визуал суваг |
|---|---|---|---|
| **Номинал (ангилал)** | `Survived` | 0 = амь үрэгдсэн, 1 = амьд үлдсэн | өнгөний туяа (hue), хэлбэр |
| | `Sex` | эрэгтэй / эмэгтэй | өнгөний туяа, панел, хэлбэр |
| | `Embarked` | C, Q, S боомт | өнгөний туяа, хэлбэр |
| **Эрэмбэт (ordinal)** | `Pclass` | 1-р, 2-р, 3-р анги | байрлал, гэрэлтэлт (lightness) |
| **Тоон (quantitative)** | `Age` | 0.4–80 нас | байрлал, урт |
| | `Fare` | £0–£512 тасалбарын үнэ | байрлал (лог масштаб), урт |
| | `SibSp`, `Parch` → `FamilySize` | 1–11 хүн | байрлал, урт, хэмжээ |

Үүсгэсэн хувьсагч: **FamilySize** = SibSp + Parch + 1; **AgeGroup** = Хүүхэд (0–17), Залуус (18–39), Ахмад (40+).

## 2. График байгуулах дүрэм (бүх зурагт мөрдсөн)

1. **Харааны мэдрэхүйн эрэмбэ (Cleveland & McGill).** Хамгийн чухал харьцуулалтыг **нийтлэг суурь дээрх байрлал** → урт → өнцөг → талбай → өнгөөр кодолно.
2. **“Start with gray” (Schwabish).** Бүх зурагт **цэнхэр = амьд үлдсэн**, **саарал = амь үрэгдсэн**. Саарал нь контекст болж, анхаарлыг амьд үлдэгсэд рүү чиглүүлнэ. Өнгө нэг notebook-д нэг л утгатай.
3. **Өнгөний төрлийг үүргээр сонгоно.** Ангилал → ялгаатай туяа, хэмжээ → нэг туяаны цайвараас бараан (sequential). Солонгон палитр хэрэглэхгүй. Цэнхэр–саарал хос нь өнгө ялгах чадвар сул хүнд ч ялгагдана (гэрэлтэлтийн ялгаатай).
4. **Хог дүрслэлийг багасгах (declutter).** Дээд/баруун хүрээ, 3D, сүүдэр байхгүй; сүлжээ шугам цайвар.
5. **Графикт текст хамгийн бага.** Зурган дээр зөвхөн **идэвхтэй гарчиг** (дүгнэлт), тайлбар (legend) ба цөөн гол тоо үлдэнэ. Кодлол, n, хассан мөр зэрэг дэлгэрэнгүйг графикийн доорх текстэд бичнэ.
6. **Үнэн зөв масштаб.** Баганыг 0-ээс эхлүүлнэ; хазайлт ихтэй `Fare`-г **лог масштабаар**; жижиг олон графикт **ижил тэнхлэг**.
7. **Давхцал (over-plotting).** Ил тод байдал, тархаалт (jitter), дэвсгэр өнгөний хүрээгээр шийднэ; jitter нэмсэн бол заавал тэмдэглэнэ.
8. **Эх сурвалж:** Titanic өгөгдөл (seaborn-data, n = 891). Хассан мөр, n < 10 бүлгийг аргачлал бүрийн *Тэмдэглэл*-д заана.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as mtick
from matplotlib.colors import LinearSegmentedColormap, ListedColormap, BoundaryNorm, Normalize
from matplotlib.lines import Line2D
from matplotlib.patches import Patch, Rectangle
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# ---- Нэгдсэн дизайны параметр ----
SURFACE = '#fcfcfb'; INK = '#0b0b0b'; INK2 = '#52514e'; MUTED = '#8a8984'; GRID = '#e6e5e1'
C_SURV = '#2a78d6'          # амьд үлдсэн — цэнхэр (онцлох өнгө)
C_DIED = '#a3a29c'          # амь үрэгдсэн — саарал ("start with gray")
PAL = {'Амьд үлдсэн': C_SURV, 'Амь үрэгдсэн': C_DIED}
# Хэмжээ (sequential) — нэг туяа: цэнхэр (амьд үлдсэн хувь), ягаан-хөх (бусад хэмжигдэхүүн)
BLUES = LinearSegmentedColormap.from_list('blues', ['#e3eefc', '#86b6ef', '#2a78d6', '#184f95', '#0d366b'])
VIOLET = LinearSegmentedColormap.from_list('violet', ['#eeebfc', '#a99ef0', '#6250d6', '#3a2c9c', '#1f175c'])

plt.rcParams.update({
    'figure.facecolor': SURFACE, 'axes.facecolor': SURFACE, 'savefig.facecolor': SURFACE,
    'font.family': 'DejaVu Sans', 'font.size': 10.5,
    'text.color': INK, 'axes.labelcolor': INK2, 'xtick.color': INK2, 'ytick.color': INK2,
    'axes.edgecolor': MUTED, 'axes.linewidth': 0.8,
    'axes.spines.top': False, 'axes.spines.right': False,
    'grid.color': GRID, 'grid.linewidth': 0.8,
    'axes.titlesize': 11.5, 'axes.titleweight': 'bold', 'axes.titlelocation': 'left',
    'legend.frameon': False, 'figure.dpi': 110,
})

def header(fig, title, gap=0.75):
    # Зөвхөн идэвхтэй гарчиг (дүгнэлт). Бусад тайлбар нь доорх текстэд.
    h = fig.get_figheight()
    fig.text(0.012, 1 - 0.12 / h, title, ha='left', va='top', fontsize=14, fontweight='bold', color=INK)
    fig.subplots_adjust(top=1 - gap / h)

def status_legend(fig, handles=None):
    # Тайлбарыг гарчгийн мөрөнд, баруун талд
    h = handles or [Patch(color=C_SURV, label='Амьд үлдсэн'), Patch(color=C_DIED, label='Амь үрэгдсэн')]
    return fig.legend(handles=h, ncol=len(h), fontsize=10, handlelength=1.2, loc='upper right',
                      bbox_to_anchor=(0.99, 1 - 0.06 / fig.get_figheight()))
```

```python
try:
    df = sns.load_dataset('titanic')
except Exception:
    df = pd.read_csv('titanic.csv')          # интернетгүй үед хажуудах файлаас

df['status'] = df['survived'].map({1: 'Амьд үлдсэн', 0: 'Амь үрэгдсэн'})
df['sex_mn'] = df['sex'].map({'female': 'Эмэгтэй', 'male': 'Эрэгтэй'})
df['FamilySize'] = df['sibsp'] + df['parch'] + 1
df['AgeGroup'] = pd.cut(df['age'], [0, 18, 40, np.inf], right=False,
                        labels=['Хүүхэд (0–17)', 'Залуус (18–39)', 'Ахмад (40+)'])
OVERALL = df['survived'].mean()

# Хувьсагчдын бүтцийн хүснэгт
vars_tbl = pd.DataFrame({
    'Хувьсагч': ['survived', 'sex', 'embarked', 'pclass', 'age', 'fare', 'FamilySize'],
    'Түвшин':   ['Номинал', 'Номинал', 'Номинал', 'Эрэмбэт', 'Тоон', 'Тоон', 'Тоон (дискрет)'],
})
vars_tbl['Ялгаатай утгын тоо'] = [df[v].nunique() for v in vars_tbl['Хувьсагч']]
vars_tbl['Дутуу утга'] = [df[v].isna().sum() for v in vars_tbl['Хувьсагч']]
print(f'Зорчигч: {len(df)},  амьд үлдсэн: {df.survived.sum()} ({OVERALL:.1%})')
vars_tbl
```

## Аргачлал 1: Mosaic plot — шатлалтай үүрэн дүрслэл (Nested visualization)

**Онолын ангилал:** Tominski & Schumann §3.2.6 *Nested visualization / Dimensional stacking* — дэлгэцийн талбайг хувьсагч бүрээр **дараалан, шатлан** хуваана.

**Визуал буулгалт**

| Элемент | Кодлол |
|---|---|
| Тэмдэглэгээ | 2D талбай — үүрлэсэн тэгш өнцөгт нүд |
| 1-р түвшин (босоо хуваалт) | `Pclass` → мөрийн **өндөр** ∝ ангийн зорчигчийн тоо |
| 2-р түвшин (хэвтээ хуваалт) | `Sex` → нүдний **өргөн** ∝ ангийн доторх хүйсийн хувь |
| 3-р түвшин (босоо хуваалт) | `Survived` → нүдний доод **цэнхэр** хэсэг = амьд үлдсэн хувь, дээд **саарал** = амь үрэгдсэн |
| Талбай | нүдний талбай ∝ зорчигчийн тоо (давтамж) |

*Тэмдэглэл:* нүдэн дээрх хувь = тухайн бүлгийн амьд үлдсэн хувь. Ангиудын зорчигч: 1-р 216 (24%), 2-р 184 (21%), 3-р 491 (55%).

```python
N = len(df)
gy, gx, gs = 0.022, 0.012, 0.006           # мөр, багана, нүдний доторх зай
avail_h = 1 - 2 * gy

fig, ax = plt.subplots(figsize=(11, 7.2))
y_top = 1.0
for c in [1, 2, 3]:
    dc = df[df.pclass == c]
    h = avail_h * len(dc) / N
    y0 = y_top - h
    x0 = 0.0
    for s, s_mn in [('female', 'Эмэгтэй'), ('male', 'Эрэгтэй')]:
        dcs = dc[dc.sex == s]
        w = (1 - gx) * len(dcs) / len(dc)
        p = dcs.survived.mean()
        hs = h * p
        ax.add_patch(Rectangle((x0, y0), w, hs - gs / 2, color=C_SURV, lw=0))           # амьд — доод суурь дээр
        ax.add_patch(Rectangle((x0, y0 + hs + gs / 2), w, h - hs - gs / 2, color=C_DIED, lw=0))
        lbl = f'{s_mn}  {p:.0%}'
        ly = y0 + 0.012 if p >= 0.3 else y0 + h - 0.012                                  # шошгыг өгөгдөл дарахгүй газар
        ax.text(x0 + 0.01, ly, lbl, va='bottom' if p >= 0.3 else 'top', fontsize=10.5, fontweight='bold',
                color='white' if p >= 0.3 else INK, zorder=5)
        x0 += w + gx
    ax.text(-0.015, y0 + h / 2, f'{c}-р анги', ha='right', va='center', fontsize=11.5, fontweight='bold')
    y_top = y0 - gy

ax.set_xlim(0, 1); ax.set_ylim(0, 1); ax.axis('off')
status_legend(fig)
header(fig, '3-р ангийн эрэгтэйчүүд: хамгийн том бүлэг, 14% амьд')
fig.subplots_adjust(left=0.11, right=0.99, bottom=0.02)
plt.show()
```

**Дүрсэлсэн өгөгдлийн тайлбар**
- **1-р ангийн эмэгтэйчүүдийн** нүдний бараг бүх талбай цэнхэр (**97%** амьд үлдсэн, n = 94), **2-р ангийнх** мөн адил (**92%**).
- **3-р ангийн эрэгтэйчүүдийн** нүд зураг дээрх **хамгийн том** талбай (347 хүн, нийт зорчигчийн 39%) бөгөөд саарал давамгайлсан — ердөө **14%** амьд үлдсэн. Энэ нэг нүд л 300 орчим амь үрэгдэгсдийг (нийт 549-ийн 55%) илэрхийлнэ.
- Нүдний өргөн ангиас хамаарч өөр байгаа нь **анги ба хүйс хоорондоо хамааралтай** гэдгийг харуулна: 1-р ангид эмэгтэйчүүд 44%, 3-р ангид ердөө 29%.
- 3-р ангийн эмэгтэйчүүд (50%) 1-р ангийн эрэгтэйчүүдээс (37%) илүү амьд үлдсэн — **хүйс ангиас илүү хүчтэй** нөлөөлсөн.

**Аргачлалын давуу тал**
- Ангиллын олон хувьсагчийн огтлолцол ба **нөхцөлт давтамжийг** (conditional frequencies) талбайн харьцаагаар нэг зурагт харуулна. Өгөгдлийн цэг давхцах (over-plotting) асуудал огт гардаггүй.
- Нэг зурагт **хувь (цэнхэр өндөр) ба жин (нүдний талбай)** хоёуланг нь харуулна: хувь өндөр ч жижиг бүлэг, хувь бага ч асар том бүлгийг ялгаж өгнө. Ангиллын өгөгдлийн хамаарлыг харуулах сонгодог арга учир Tominski, Schumann нар Титаникийн өгөгдлөөр жишээ болгон авч үздэг (Зураг 3.27).

*Хязгаарлалт:* 4-өөс олон хувьсагч үүрлэхэд нүд хэт жижгэрч уншигдахгүй болно; тоон хувьсагчийг эхлээд ангилалд хувиргах шаардлагатай.

## Аргачлал 2: Parallel Coordinates Plot (PCP) — зэрэгцээ координат

**Онолын ангилал:** Tominski & Schumann §3.2.3 *Polyline-based visualization*; Schwabish — *Relationship charts*.

**Визуал буулгалт**

| Элемент | Кодлол |
|---|---|
| Тэмдэглэгээ | 1D шугам (polyline) — нэг зорчигч = нэг тасархай шугам |
| Байрлал (5 зэрэгцээ босоо тэнхлэг) | `Pclass` → `Fare` (лог) → `Age` → `FamilySize` → `Survived` |
| Өнгөний туяа | `Survived`: амь үрэгдсэн = **саарал** (эхэлж, доор зурна), амьд үлдсэн = **цэнхэр** (дээр нь) |
| Ил тод байдал | 700+ шугамын давхцлыг багасгах |

*Тэмдэглэл:* тэнхлэг бүрийн дээд тал = дээд анги / өндөр үнэ / их нас / том гэр бүл / амьд үлдсэн. Анги, гэр бүл, хувь заяанд jitter нэмсэн; нас тодорхойгүй 177 мөрийг хассан (n = 714: амьд 290, үрэгдсэн 424); £4-өөс доош үнийг £4 гэж харуулсан.

```python
d2 = df.dropna(subset=['age']).copy()
rng = np.random.default_rng(3)
fare_l = np.log10(d2.fare.clip(lower=4))
axes_def = [   # (нэр, утга, доод, дээд, тэмдэглэгээ)   — дээд тал = "их/сайн"
    ('Анги',            (4 - d2.pclass) + rng.uniform(-.12, .12, len(d2)), 0.7, 3.3, {3: '1-р', 2: '2-р', 1: '3-р'}),
    ('Үнэ',             fare_l, np.log10(4), np.log10(520), {np.log10(v): f'£{v}' for v in [10, 100]}),
    ('Нас',             d2.age, 0, 80, {v: str(v) for v in [0, 40, 80]}),
    ('Гэр бүл',         d2.FamilySize + rng.uniform(-.15, .15, len(d2)), 0.6, 11.4, {1: '1', 11: '11'}),
    ('Хувь заяа',       d2.survived + rng.uniform(-.06, .06, len(d2)), -0.25, 1.25, {1: 'Амьд', 0: 'Үрэгдсэн'}),
]
Y = np.column_stack([(v - lo) / (hi - lo) for _, v, lo, hi, _ in axes_def])
xs = np.arange(len(axes_def))
surv = d2.survived.values == 1

fig, ax = plt.subplots(figsize=(12, 6.6))
ax.plot(xs, Y[~surv].T, color=C_DIED, alpha=0.13, lw=0.7, zorder=1)
ax.plot(xs, Y[surv].T, color=C_SURV, alpha=0.20, lw=0.8, zorder=2)
for i, (name, _, lo, hi, ticks) in enumerate(axes_def):
    ax.axvline(i, color=INK2, lw=0.9, zorder=3)
    for tv, tl in ticks.items():
        ax.text(i - 0.035, (tv - lo) / (hi - lo), tl, ha='right', va='center', fontsize=8.5, color=INK2, zorder=5,
                bbox=dict(boxstyle='round,pad=0.12', fc=SURFACE, ec='none', alpha=0.85))
ax.set_xticks(xs, [n for n, *_ in axes_def], fontsize=10.5)
ax.tick_params(axis='x', length=0)
ax.set_yticks([]); ax.spines[['left', 'bottom']].set_visible(False)
ax.set_xlim(-0.35, len(xs) - 0.75); ax.set_ylim(-0.03, 1.03)

status_legend(fig, [Line2D([], [], color=C_SURV, lw=2.5, label='Амьд үлдсэн'),
                   Line2D([], [], color=C_DIED, lw=2.5, label='Амь үрэгдсэн')])
header(fig, 'Дээд анги, өндөр үнэ → амьд үлдэх; 3-р анги, хямд үнэ → амь үрэгдэх', gap=0.9)
fig.subplots_adjust(left=0.04, right=0.98, bottom=0.07)
plt.show()
```

```python
# Шугамын урсгалыг тоогоор баталгаажуулах: анги × үнийн түвшин бүрийн амьд үлдсэн хувь
d2['Үнийн түвшин'] = pd.cut(d2.fare, [-1, 10, 30, 100, 600], labels=['≤ £10', '£10–30', '£30–100', '> £100'])
d2.pivot_table('survived', 'pclass', 'Үнийн түвшин', aggfunc='mean', observed=False).round(2)
```

**Дүрсэлсэн өгөгдлийн тайлбар**
- **Анги → Үнэ** хооронд шугамууд бараг **хэвтээ, зэрэгцээ** урсна: 1-р анги дээд үнэ рүү, 3-р анги доод үнэ рүү. Энэ бол хүчтэй эерэг хамаарал (анги өндөр ↔ үнэ өндөр).
- 1-р ангиас ба **£30-аас дээш** үнээс эхэлсэн **цэнхэр** шугамууд **Нас** тэнхлэгийн бүх мужийг (0–80) дайрч “Амьд үлдсэн” рүү очдог — өндөр анги/үнэтэй хүмүүсийн хувьд нас бараг нөлөөгүй.
- 3-р ангиас ба **≤ £10** үнээс гарсан шугамууд бөөнөөрөө **саарал** бөгөөд “Амь үрэгдсэн” рүү бөөгнөрнө. Хүснэгтээр ≤ £10 үнэтэй 3-р ангийнхны ердөө **20%** орчим нь амьд үлдсэн.
- **Гэр бүлийн хэмжээ** тэнхлэгийн **дээд хэсгээс** (5+ хүн) “Амьд үлдсэн” рүү очих цэнхэр шугам маш цөөн — том гэр бүл өндөр эрсдэлтэй байсан.

**Аргачлалын давуу тал**
- Scatterplot зөвхөн 2 хувьсагчийг харуулдаг бол PCP нь **олон тэнхлэгийг нэгэн зэрэг** харуулж, 5 хэмжээст өгөгдлийг 2D хавтгайд буулгана.
- Хувьсагч хоорондын **корреляц** (зэрэгцээ шугам = эерэг, огтлолцох = сөрөг), зорчигчдын **профиль хэв маяг** ба **гаж утгыг** (жишээ нь £500-ийн тасалбар) нэг дэлгэц дээр шинжилнэ.
- Schwabish-ийн **“start with gray”** зарчмаар амь үрэгдэгсдийг саарал контекст болгосноор цэнхэр шугамын урсгал “спагетти” дунд төөрөхгүй.

*Хязгаарлалт:* тэнхлэгийн **дараалал** үр дүнд нөлөөлнө (зөвхөн хөрш тэнхлэгийн хамаарал харагдана); ангиллын тэнхлэгт шугам нэг цэгт бөөгнөрөх тул jitter хэрэгтэй.

## Аргачлал 3: Scatter Plot Matrix (SPLOM) — хосолсон 2D матрицын дүрслэл

**Онолын ангилал:** Tominski & Schumann §3.2.2 *Combined bivariate visualization*; Schwabish — *Visual perception rankings*.

**Визуал буулгалт**

| Элемент | Кодлол |
|---|---|
| Тэмдэглэгээ | 0D цэг, *m × m* торлолтой матриц (энд *m* = 3) |
| Байрлал (X, Y) | `Age`, `Fare` (лог), `FamilySize` хувьсагчдын хос бүр |
| Өнгөний туяа | `Survived`: цэнхэр / саарал |
| Хэлбэр (shape) | `Pclass`: ● 1-р, ■ 2-р, ▲ 3-р анги |
| Диагональ | хувьсагч тус бүрийн **гистограмм** (амьд үлдсэн нь нийтлэг суурь дээр, давхарласан) |

*Тэмдэглэл:* гэр бүлийн хэмжээнд ±0.22 jitter нэмсэн; нас тодорхойгүй 177 мөрийг хассан (n = 714).

```python
d3 = df.dropna(subset=['age']).copy()
rng = np.random.default_rng(7)
d3['fare_l'] = np.log10(d3.fare.clip(lower=4))
d3['fam_j'] = d3.FamilySize + rng.uniform(-0.22, 0.22, len(d3))
V = [('age', 'Нас (жил)', np.arange(0, 85, 5)),
     ('fare_l', 'Үнэ (£, лог)', np.linspace(np.log10(4), np.log10(520), 18)),
     ('fam_j', 'Гэр бүл', np.arange(0.5, 12, 1))]
MARK = {1: 'o', 2: 's', 3: '^'}
fare_ticks = [np.log10(v) for v in [5, 10, 30, 100, 500]]
fare_lbls = ['£5', '£10', '£30', '£100', '£500']

m = len(V)
fig, axs = plt.subplots(m, m, figsize=(11, 10.5))
for i, (vy, ly, by) in enumerate(V):
    for j, (vx, lx, bx) in enumerate(V):
        ax = axs[i, j]
        if j > i:
            ax.axis('off'); continue
        if i == j:   # диагональ: овоолсон гистограмм — амьд үлдсэн нь суурь дээр
            vals = [d3.loc[d3.survived == 1, vx], d3.loc[d3.survived == 0, vx]]
            ax.hist(vals, bins=bx, stacked=True, color=[C_SURV, C_DIED], edgecolor=SURFACE, linewidth=0.6)
            ax.set_yticks([]); ax.spines['left'].set_visible(False)
        else:
            for st, col, z in [(0, C_DIED, 1), (1, C_SURV, 2)]:
                for c, mk in MARK.items():
                    s = d3[(d3.survived == st) & (d3.pclass == c)]
                    ax.scatter(s[vx], s[vy], marker=mk, s=20, color=col, alpha=0.65,
                               edgecolor=SURFACE, linewidth=0.4, zorder=z)
            ax.grid(True); ax.set_axisbelow(True)
        if vx == 'fare_l': ax.set_xticks(fare_ticks, fare_lbls)
        if vy == 'fare_l' and i != j: ax.set_yticks(fare_ticks, fare_lbls)
        if i == m - 1: ax.set_xlabel(lx)
        else: ax.set_xticklabels([])
        if j == 0 and i != 0: ax.set_ylabel(ly)
        if j != 0 and i != j: ax.set_yticklabels([])
axs[2, 2].set_xticks([1, 3, 5, 7, 9, 11]); axs[2, 0].set_yticks([1, 3, 5, 7])

h_col = [Patch(color=C_SURV, label='Амьд үлдсэн'), Patch(color=C_DIED, label='Амь үрэгдсэн')]
h_shp = [Line2D([], [], marker=mk, ls='', color=INK2, markersize=7, label=f'{c}-р анги') for c, mk in MARK.items()]
fig.legend(handles=h_col + h_shp, loc='upper left', bbox_to_anchor=(0.62, 0.82), fontsize=10.5, labelspacing=0.8)
header(fig, 'Үнэ ба анги хамгийн сайн салгагч; нас ганцаараа сул')
fig.subplots_adjust(left=0.08, right=0.98, bottom=0.06, hspace=0.12, wspace=0.12)
plt.show()
```

```python
# Өндөр үнэтэй (> £100) зорчигчид: амьд үлдсэн хувь нас ба хүйсээр
hi = df[df.fare > 100]
pd.concat({
    'Насаар': hi.groupby(pd.cut(hi.age, [0, 30, 80], labels=['≤ 30 нас', '> 30 нас']), observed=True).survived.agg(['size', 'mean']),
    'Хүйсээр': hi.groupby('sex_mn').survived.agg(['size', 'mean']),
}).rename(columns={'size': 'n', 'mean': 'амьд үлдсэн хувь'}).round(2)
```

**Дүрсэлсэн өгөгдлийн тайлбар**
- **Үнэ × Нас** хос дээр **£100-аас дээш** үнийн бүсэд (ихэвчлэн ● 1-р анги) цэнхэр цэг давамгайлна. Гэхдээ “өндөр үнэтэй залуус бараг бүгд амьд үлдсэн” гэж хэлэх нь хэтрүүлэг: хүснэгтээр > £100 тасалбартай **30-аас доош** насныхны **68%**, **30-аас дээш** насныхны **81%** амьд үлдсэн. Энэ бүсэд ялгааг **нас биш хүйс** тодорхойлдог (эмэгтэй **94%**, эрэгтэй **37%**).
- **£10-аас доош** бүс ▲ 3-р ангийн гурвалжнуудаар дүүрэн, голдуу саарал. Цэгийн хэлбэр ба өнгийг хамт уншихад “**хямд тасалбар = 3-р анги = өндөр эрсдэл**” гэсэн гурван хувьсагчийн хамаарал нэг самбарт харагдана.
- **Нас** диагональд: цэнхэр баганууд зөвхөн **0–10 насанд** саарлаасаа давна (хүүхдүүд). Бусад насанд хоёр өнгийн харьцаа бараг тогтмол — нас ганцаараа сул таамаглагч.
- **Гэр бүл × Нас:** 5+ хүнтэй гэр бүлийн цэгүүд (дээд хэсэг) бараг бүгд ▲ саарал, бага насныхан — 3-р ангийн олон хүүхэдтэй өрхүүд.

**Аргачлалын давуу тал**
- Cleveland & McGill-ийн эрэмбээр **нийтлэг масштаб дээрх байрлал** нь тоон утгыг хамгийн нарийн уншуулдаг суваг. SPLOM нь бүх хос хувьсагчийг энэ сувгаар **системтэйгээр** харуулна.
- Нэг 2D графикт харагдсан хамаарал бусад самбарт хэрхэн илэрч байгааг харьцуулж, **хуурамч хамаарлыг** (жишээ нь “залуу + өндөр үнэ” гэсэн дүгнэлтийг хүйсээр тайлбарлах) олон хувьсагчийн түвшинд шалгана. Интерактив **brushing & linking**-тэй хослуулбал нэг самбарт сонгосон цэгүүд бусад бүх самбарт тодорно.

*Хязгаарлалт:* *m* өсөхөд самбарын тоо *m²*-аар өсч, тус бүр жижгэрнэ; хэлбэр + өнгө хоёр сувгийг зэрэг кодлоход нягт бүсэд хэлбэр ялгагдахад хэцүү.

## Аргачлал 4: Heatmap — өнгөт матрицын (хүснэгтэд суурилсан) дүрслэл

**Онолын ангилал:** Tominski & Schumann §3.2.1 *Table-based visualization*; Schwabish — *Heatmap*.

**Визуал буулгалт**

| Элемент | Кодлол |
|---|---|
| Тэмдэглэгээ | 2D хүснэгтийн нүд (grid cell) |
| Мөр (6) | `Pclass` × `Sex` хослол |
| Багана (3) | `AgeGroup`: Хүүхэд (0–17), Залуус (18–39), Ахмад (40+) |
| Өнгөний гэрэлтэлт/ханалт | нүдний **амьд үлдсэн хувь** (0% цайвар → 100% бараан цэнхэр, нэг туяа) |
| Текст | хувийг нүдэнд давхар бичнэ |

*Тэмдэглэл:* нас тодорхойгүй 177 мөрийг хассан. Бүлгийн хэмжээ: 1-р ангийн эмэгтэй хүүхэд n = 8, эрэгтэй хүүхэд n = 4 (n < 10 — хувийг болгоомжтой унш); бусад нүд n = 10–175.

```python
d4 = df.dropna(subset=['age'])
rate = d4.pivot_table('survived', ['pclass', 'sex_mn'], 'AgeGroup', aggfunc='mean', observed=False)
cnt = d4.pivot_table('survived', ['pclass', 'sex_mn'], 'AgeGroup', aggfunc='size', observed=False)
rate.index = [f'{c}-р анги · {s}' for c, s in rate.index]

fig, ax = plt.subplots(figsize=(9.5, 7))
im = ax.imshow(rate.values, cmap=BLUES, vmin=0, vmax=1, aspect='auto')
for i in range(rate.shape[0]):
    for j in range(rate.shape[1]):
        v, n = rate.values[i, j], cnt.values[i, j]
        col = 'white' if v > 0.55 else INK
        ax.text(j, i, f'{v:.0%}', ha='center', va='center', fontsize=14, fontweight='bold', color=col)
# нүд хооронд дэвсгэр өнгийн зай
ax.set_xticks(np.arange(-.5, rate.shape[1]), minor=True); ax.set_yticks(np.arange(-.5, rate.shape[0]), minor=True)
ax.grid(which='minor', color=SURFACE, linewidth=3); ax.tick_params(which='both', length=0)
ax.set_xticks(range(rate.shape[1]), rate.columns, fontsize=10.5); ax.xaxis.tick_top()
ax.set_yticks(range(rate.shape[0]), rate.index, fontsize=10.5)
for sp in ax.spines.values(): sp.set_visible(False)
for y in [1.5, 3.5]: ax.axhline(y, color=SURFACE, lw=8)        # анги хооронд илүү зай
cb = fig.colorbar(im, ax=ax, shrink=0.6, pad=0.03, ticks=[0, 0.5, 1], format=mtick.PercentFormatter(1.0, decimals=0))
cb.outline.set_visible(False); cb.ax.tick_params(length=0)
header(fig, 'Хамгийн “халуун” нүд: 1–2-р ангийн эмэгтэйчүүд ба хүүхдүүд', gap=1.0)
fig.subplots_adjust(left=0.22, bottom=0.03)
plt.show()
```

**Дүрсэлсэн өгөгдлийн тайлбар**
- **Хамгийн бараан (халуун) нүднүүд:** 1-р ангийн эмэгтэй залуус **98%**, ахмад **96%**; 2-р ангийн эмэгтэй хүүхэд **100%**, залуус **91%**; 1-р ангийн эрэгтэй хүүхэд **100%** (n = 4). Эмэгтэй 1–2-р анги болон 1–2-р ангийн хүүхдүүдийн нүд **82–100%** мужид байна.
- **Хүүхэд** багана **ангиас хүчтэй хамаардаг**: 2-р ангийн хөвгүүд **82%** бол 3-р ангийн хөвгүүд ердөө **23%**, 3-р ангийн охид **54%**. Иймд “хүүхдүүд амьд үлдсэн” гэдэг дүгнэлт зөвхөн 1–2-р ангид хүчинтэй.
- **Хамгийн цайвар (хүйтэн) нүднүүд:** 2-р ангийн эрэгтэй залуус **6%**, 3-р ангийн эрэгтэй ахмад **6%**, 2-р ангийн эрэгтэй ахмад **10%**, 3-р ангийн эмэгтэй ахмад **10%**.
- Эрэгтэй мөрүүдэд хүүхдээс залуус руу өнгө **огцом цайрдаг** (2-р анги: 82% → 6%) — “эмэгтэйчүүд, хүүхдийг эхэлж” зарчмын нөлөө.

**Аргачлалын давуу тал**
- 18 тоог агуулсан хүснэгтийг уншихад шаардагдах **танин мэдэхүйн ачааллыг** (cognitive effort) бууруулна. Өнгөний ялгааг хүн **урьдчилан мэдрэх (preattentive)** үйл явцаар 200–250 мс дотор ялгадаг тул хамгийн өндөр, хамгийн бага **“халуун цэгүүд”** (hotspots) шууд харагдана.
- Мөр ба баганад өөр өөр ангиллын хувьсагчийг **хослуулж** (энд анги × хүйс) 3 хувьсагчийн нөхцөлт утгыг нягт хүснэгтэд багтаана. Тоог давхар бичсэн тул өнгөний нарийвчлал бага гэсэн сул тал нөхөгдөнө.

*Хязгаарлалт:* өнгөний гэрэлтэлт нь урт/байрлалаас **нарийвчлал муутай** суваг; нүд бүрийн **түүврийн хэмжээ** (n) харагдахгүй тул n-ийг тусад нь бичих хэрэгтэй.

## Аргачлал 5: Small Multiples — олон жижиг дүрслэлийн сүлжээ (Faceted grid)

**Онолын ангилал:** Tominski & Schumann §3.3.1 *Faceting*; Schwabish — *Small multiples* зөвлөмж.

**Визуал буулгалт**

| Элемент | Кодлол |
|---|---|
| Сүлжээ | 2 × 3 панел: мөр = `Sex`, багана = `Pclass` |
| Байрлал (X) | `Age` — 5 жилийн интервал |
| Урт / өндөр (Y) | зорчигчийн **тоо** |
| Өнгөний туяа | `Survived`: **цэнхэрийг нийтлэг суурь (0) дээр**, саарлыг дээр нь овоолно |
| Тэнхлэг | бүх панел **ижил X, Y масштаб** |

*Тэмдэглэл:* панел бүрийн баруун дээд хувь = тухайн бүлгийн нийт амьд үлдсэн хувь (насгүй мөрийг оруулсан). Гистограммд нас тодорхойгүй 177 мөрийг хассан.

```python
d5 = df.dropna(subset=['age'])
bins = np.arange(0, 85, 5)
fig, axs = plt.subplots(2, 3, figsize=(13, 7.6), sharex=True, sharey=True)
for i, (s, s_mn) in enumerate([('female', 'Эмэгтэй'), ('male', 'Эрэгтэй')]):
    for j, c in enumerate([1, 2, 3]):
        ax = axs[i, j]
        sub = d5[(d5.sex == s) & (d5.pclass == c)]
        ax.hist([sub.loc[sub.survived == 1, 'age'], sub.loc[sub.survived == 0, 'age']], bins=bins,
                stacked=True, color=[C_SURV, C_DIED], edgecolor=SURFACE, linewidth=1.0, zorder=2)
        p_all = df[(df.sex == s) & (df.pclass == c)].survived.mean()
        ax.set_title(f'{s_mn} · {c}-р анги', fontsize=11)
        ax.text(0.98, 0.95, f'{p_all:.0%}', transform=ax.transAxes, ha='right', va='top',
                fontsize=13, fontweight='bold', color=C_SURV)
        ax.grid(axis='y'); ax.set_axisbelow(True)
        if i == 1: ax.set_xlabel('Нас (жил)')
        if j == 0: ax.set_ylabel('Зорчигчийн тоо')
axs[1, 0].set_xticks(range(0, 81, 20))
status_legend(fig)
header(fig, 'Эрэгтэйчүүдээс голдуу хүүхдүүд, 1–2-р ангийн эмэгтэйчүүд бараг бүгд амьд үлдсэн', gap=0.85)
fig.subplots_adjust(left=0.06, right=0.98, bottom=0.08, hspace=0.28, wspace=0.08)
plt.show()
```

**Дүрсэлсэн өгөгдлийн тайлбар**
- **Эмэгтэй · 1-р, 2-р анги** панелд **наснаас үл хамааран** бараг бүх багана бүхэлдээ цэнхэр (97%, 92%).
- **Эрэгтэй** мөрөнд цэнхэр хэсэг зөвхөн **18-аас доош** насанд (хүүхдүүд) өндөр: 1–2-р ангийн 12 хүртэлх насны 12 хөвгүүд **бүгд** амьд үлдсэн. Харин 18-аас дээш насанд баганууд бараг бүхэлдээ саарал: насанд хүрсэн эрэгтэйчүүдийн 1-р ангид **37%**, 2-р ангид **7%**, 3-р ангид **13%**.
- **Эрэгтэй · 3-р анги** — бүх панелийн дотор **хамгийн өндөр** баганууд (20–30 насны залуус) бөгөөд бараг бүхэлдээ саарал. Энэ бол амь үрэгдэгсдийн гол бүлэг.
- **Эмэгтэй · 3-р анги** панел цорын ганц “холимог” эмэгтэй панел (50%): цэнхэр/саарал харьцаа бүх насанд ойролцоо — 3-р ангийн эмэгтэйчүүдэд нас биш **анги** шийдвэрлэх хүчин зүйл байжээ.

**Аргачлалын давуу тал**
- Олон бүлгийг нэг график дээр давхарлан **“спагетти” / визуал эмх замбараагүй байдал** үүсгэхийн оронд, **ижил суурь шугам ба масштабтай** жижиг графикуудад хуваана. Нүд панелуудыг шууд харьцуулж хамгийн нарийн харьцуулалт хийнэ.
- Панелийн сүлжээ өөрөө 2 хувьсагчийг (хүйс, анги) кодолдог тул панел доторх график энгийн хэвээр үлдэж, нийтдээ **4 хувьсагч** (хүйс, анги, нас, хувь заяа) уншигдана. Амьд үлдэгсдийг **нийтлэг суурь (0)** дээр байрлуулсан тул тэдгээрийн өндрийг панел хооронд байрлалаар зөв харьцуулна.

*Хязгаарлалт:* панелийн тоо хурдан өсдөг (энд 6); жижиг панелд нарийн утга уншигдахгүй тул гол хувийг шууд шошголсон.

## Аргачлал 6: Glyph-based visualization — Star plot (одон глиф)

**Онолын ангилал:** Tominski & Schumann §3.2.4 *Glyph-based visualization* (Chernoff faces, star plots); Schwabish — *Radar charts*.

**Визуал буулгалт**

| Элемент | Кодлол |
|---|---|
| Тэмдэглэгээ | 2D биеэ даасан олон өнцөгт дүрс (glyph) — **нэг бүлгийн профиль = нэг од** |
| Бүлэг (12 глиф) | `Pclass` (мөр) × `Sex` × `Survived` (багана) |
| Цацрагийн урт (4 тэнхлэг) | **Нас** (медиан), **Тасалбарын үнэ** (медиан, лог), **Анги** (1-р = хамгийн урт), **Гэр бүлийн хэмжээ** (дундаж) |
| Өнгөний туяа | `Survived`: цэнхэр / саарал |
| Лавлах дүрс | тасархай шугам = нийт зорчигчийн профиль |

Тэнхлэг бүрийг 12 бүлгийн утгын хамгийн бага → хамгийн их мужаар 0.15–1 хооронд нормчилсон. Тэнхлэгийн нэрийг зөвхөн эхний глифт бичсэн — бусад нь ижил байрлалтай (дээр Нас, баруун Гэр бүл, доор Анги, зүүн Үнэ).

```python
g = (df.groupby(['pclass', 'sex_mn', 'survived'])
       .agg(age=('age', 'median'), fare=('fare', 'median'), fam=('FamilySize', 'mean'), n=('survived', 'size'))
       .reset_index())
g['cls'] = 4 - g.pclass                        # 1-р анги = 3 (урт цацраг)
g['fare_l'] = np.log10(g.fare)
AX = [('age', 'Нас'), ('fare_l', 'Үнэ'), ('cls', 'Анги'), ('fam', 'Гэр бүл')]
lo = {k: g[k].min() for k, _ in AX}; hi = {k: g[k].max() for k, _ in AX}
lo['cls'], hi['cls'] = 0, 3
norm = lambda k, v: 0.15 + 0.85 * (v - lo[k]) / (hi[k] - lo[k])
ref = {'age': df.age.median(), 'fare_l': np.log10(df.fare.median()), 'cls': 4 - df.pclass.mean(), 'fam': df.FamilySize.mean()}
ang = np.linspace(0, 2 * np.pi, len(AX), endpoint=False) + np.pi / 2
close = lambda a: np.r_[a, a[:1]]

cols = [('Эмэгтэй', 1), ('Эмэгтэй', 0), ('Эрэгтэй', 1), ('Эрэгтэй', 0)]
fig, axs = plt.subplots(3, 4, figsize=(12, 9.5), subplot_kw={'projection': 'polar'})
for i, c in enumerate([1, 2, 3]):
    for j, (s, sv) in enumerate(cols):
        ax = axs[i, j]
        row = g[(g.pclass == c) & (g.sex_mn == s) & (g.survived == sv)].iloc[0]
        r = np.array([norm(k, row[k]) for k, _ in AX])
        rr = np.array([norm(k, ref[k]) for k, _ in AX])
        col = C_SURV if sv == 1 else C_DIED
        ax.plot(close(ang), close(rr), color=INK2, lw=0.9, ls=(0, (3, 2)), zorder=2)
        ax.fill(close(ang), close(r), color=col, alpha=0.45, zorder=3)
        ax.plot(close(ang), close(r), color=col, lw=2, zorder=4)
        ax.set_ylim(0, 1.05); ax.set_yticks([0.5, 1.0]); ax.set_yticklabels([])
        ax.set_xticks(ang, [lbl for _, lbl in AX] if (i, j) == (0, 0) else [], fontsize=9, color=INK2)
        ax.grid(color=GRID, lw=0.8); ax.spines['polar'].set_visible(False)
        if i == 0:
            ax.text(0.5, 1.2, f'{s} · {"амьд" if sv else "үрэгдсэн"}', transform=ax.transAxes, ha='center',
                    fontsize=11, fontweight='bold', color=col if sv else INK2)
    fig.text(0.012, 0.765 - i * 0.30, f'{c}-р\nанги', fontsize=12, fontweight='bold', va='center')

leg = [Patch(color=C_SURV, alpha=0.6, label='Амьд үлдсэн'), Patch(color=C_DIED, alpha=0.6, label='Амь үрэгдсэн'),
       Line2D([], [], color=INK2, lw=1, ls=(0, (3, 2)), label='Нийт зорчигчийн профиль')]
leg[2] = Line2D([], [], color=INK2, lw=1, ls=(0, (3, 2)), label='Нийт дундаж')
status_legend(fig, leg)
header(fig, 'Дээд анги — тэлсэн од, 3-р анги — агшсан од', gap=1.25)
fig.subplots_adjust(left=0.08, right=0.98, bottom=0.03, hspace=0.3, wspace=0.25)
plt.show()
```

```python
# Глиф бүрийн бодит утга (графикт бичээгүй дэлгэрэнгүй)
(g.assign(**{'Хувь заяа': g.survived.map({1: 'амьд', 0: 'үрэгдсэн'})})
  .rename(columns={'pclass': 'Анги', 'sex_mn': 'Хүйс', 'age': 'Нас (медиан)', 'fare': 'Үнэ £ (медиан)', 'fam': 'Гэр бүл (дундаж)'})
  [['Анги', 'Хүйс', 'Хувь заяа', 'n', 'Нас (медиан)', 'Үнэ £ (медиан)', 'Гэр бүл (дундаж)']].round(1))
```

**Дүрсэлсэн өгөгдлийн тайлбар**

*Тэмдэглэл:* глиф бүрийн бодит утга ба n-ийг доорх хүснэгтэд харуулав (графикт бичээгүй). 1-р ба 2-р ангийн амь үрэгдсэн эмэгтэйчүүд n = 3 ба 6 — болгоомжтой унш.

- **1-р анги · эмэгтэй · амьд үлдсэн** (n = 91) од нь **Анги** ба **Үнэ** цацрагийн дагуу тэлсэн, тасархай лавлах дүрснээс том, тод цэнхэр олон өнцөгт (медиан £82, 35 нас).
- **3-р анги · эрэгтэй · амь үрэгдсэн** (n = 300) од **Үнэ** (медиан £7.9), **Анги** ба **Гэр бүл** (дундаж 1.7; **77%** нь ганцаараа явсан) цацрагаараа **агшсан** саарал дүрс — лавлах дүрснээс бараг бүх талаараа жижиг.
- Сонирхолтой нь **3-р ангийн эрэгтэйчүүдийн амьд үлдсэн ба амь үрэгдсэн одод бараг ижил хэлбэртэй** (хоёулаа 25 нас, £8, гэр бүл ≈ 1.6–1.7). Энэ 4 хувьсагч тэр бүлгийн дотор хэн амьд үлдэхийг тайлбарлахгүй — бусад хүчин зүйл (хөлөг дээрх байрлал, тохиол) нөлөөлсөн байх. Хоёр глифийг хажуу хажууд тавьснаар “ялгаа байхгүй” гэдэг нь ч мэдээлэл болж харагдана.
- **2-р анги · эрэгтэй · амьд үлдсэн** (n = 17) од **Нас** цацрагаараа бараг төвд шахагдсан (медиан **3 нас**) — амьд үлдсэн 2-р ангийн эрэгтэйчүүд бараг бүгд хүүхдүүд байсныг **хэлбэрээр** нь шууд таньж болно. Харин хажуугийн амь үрэгдэгсдийн од (30 нас) огт өөр хэлбэртэй.
- **3-р анги · эмэгтэй · амь үрэгдсэн** од **Гэр бүл** цацрагаараа хамгийн урт (дундаж **3.4** хүн) — 3-р ангийн том гэр бүлийн эхчүүд.
- 1-р анги · эмэгтэй · амь үрэгдсэн (n = 3 ⚠) од хэт өндөр үнэтэй харагдаж байгаа ч гурван хүний медиан тул тайлбар хийхгүй.

**Аргачлалын давуу тал**
- Тэнхлэгт суурилсан графикаас ялгаатай нь нэг өгөгдлийн нэгжийн (энд бүлгийн) олон хувьсагчийг **нэг бүхэл биет дүрс** болгоно. Хүний хараа геометр дүрсийг хурдан таньдаг тул бүлгүүдийн **профиль (хэлбэр, footprint)** болон **онцгой глифүүдийг** (2-р ангийн амьд үлдсэн эрэгтэйчүүд) компакт зайд шууд ялгана.
- Глифүүдийг торлолд байрлуулснаар глиф ба сүлжээний байрлал хоёулаа мэдээлэл кодолно: мөр = анги, багана = хүйс × хувь заяа.

*Хязгаарлалт:* цацрагийн урт **радиал** тул нийтлэг суурьтай байрлалаас нарийвчлал муу; олон өнцөгтийн талбай нь тэнхлэгийн **дарааллаас** хамаарна. Тиймээс тоон утгыг глиф бүрийн доор бичсэн.

## Аргачлал 7: Pixel-based visualization — пикселэн олон хувьсагчийн дүрслэл (Keim)

**Онолын ангилал:** Tominski & Schumann §3.2.5 *Pixel-based visualization* (Keim & Kriegel).

**Визуал буулгалт**

| Элемент | Кодлол |
|---|---|
| Тэмдэглэгээ | 0D ганц пиксел (энд томруулсан нүд) — **нэг зорчигч = хувьсагч бүрийн цонхонд нэг пиксел** |
| Дэд цонх (4) | `Fare`, `Pclass`, `Age`, `Survived` |
| Байрлал (дараалал) | 891 зорчигчийг **`Fare`-ээр буурахаар** эрэмбэлж, 30 × 30 торлолд **мөр мөрөөр** (зүүнээс баруун, дээрээс доош) байрлуулна. Бүх цонхонд **ижил дараалал** — нэг байрлал = нэг зорчигч |
| Өнгө | тоон/эрэмбэт хувьсагч → нэг туяаны (ягаан-хөх) гэрэлтэлт; `Survived` → цэнхэр / саарал; дутуу нас → саарал |

*Тэмдэглэл:* цонх бүрийн **зүүн дээд** пиксел = хамгийн үнэтэй тасалбар, **баруун доод** = хамгийн хямд. 30 × 30 торлолын сүүлийн 9 нүд хоосон (саарал). “Нас” цонхны саарал пиксел = нас тодорхойгүй.

```python
d7 = df.sort_values('fare', ascending=False, kind='stable').reset_index(drop=True)
W = 30; H = int(np.ceil(len(d7) / W))
def grid(values):
    a = np.full(W * H, np.nan); a[:len(values)] = values
    return a.reshape(H, W)

windows = [
    ('Үнэ (эрэмбэлэх)', grid(np.log10(d7.fare.clip(lower=4))), VIOLET,
     Normalize(np.log10(4), np.log10(520)), 'fare'),
    ('Анги', grid(4 - d7.pclass), ListedColormap(['#c6bff5', '#6250d6', '#1f175c']),
     BoundaryNorm([0.5, 1.5, 2.5, 3.5], 3), 'cls'),
    ('Нас', grid(d7.age), VIOLET, Normalize(0, 80), 'age'),
    ('Хувь заяа', grid(d7.survived), ListedColormap([C_DIED, C_SURV]), BoundaryNorm([-0.5, 0.5, 1.5], 2), 'surv'),
]
fig, axs = plt.subplots(1, 4, figsize=(14, 4.3))
for ax, (title, A, cmap, nrm, key) in zip(axs, windows):
    cm = cmap.copy(); cm.set_bad('#cfcec8')                        # дутуу / хоосон пиксел
    im = ax.imshow(np.ma.masked_invalid(A), cmap=cm, norm=nrm, interpolation='nearest')
    ax.set_xticks([]); ax.set_yticks([])
    for sp in ax.spines.values(): sp.set_visible(False)
    ax.set_title(title, fontsize=11)
    cb = fig.colorbar(im, ax=ax, orientation='horizontal', fraction=0.05, pad=0.04, aspect=25)
    cb.outline.set_visible(False); cb.ax.tick_params(labelsize=8.5, length=0)
    if key == 'fare':
        tv = [10, 100]; cb.set_ticks(np.log10(tv)); cb.set_ticklabels([f'£{v}' for v in tv])
    elif key == 'cls':
        cb.set_ticks([1, 2, 3]); cb.set_ticklabels(['3-р', '2-р', '1-р'])
    elif key == 'age':
        cb.set_ticks([0, 40, 80])
    else:
        cb.set_ticks([0, 1]); cb.set_ticklabels(['Үрэгдсэн', 'Амьд'])
header(fig, 'Тасалбар хямдрах тусам цэнхэр пиксел цөөрнө', gap=0.55)
fig.subplots_adjust(left=0.02, right=0.98, bottom=0.02, wspace=0.08)
plt.show()
```

```python
# Пикселэн дүрслэлийн дүгнэлтийг тоогоор: үнийн 10 хэсэг (decile) бүрийн амьд үлдсэн хувь, 1-р ангийн хувь
d7['Үнийн decile'] = pd.qcut(d7.fare.rank(method='first', ascending=False), 10, labels=[f'{i}' for i in range(1, 11)])
(d7.groupby('Үнийн decile', observed=True)
   .agg(**{'медиан үнэ £': ('fare', 'median'), 'амьд үлдсэн': ('survived', 'mean'),
           '1-р анги': ('pclass', lambda s: (s == 1).mean()), 'нас дутуу': ('age', lambda s: s.isna().mean())})
   .round(2).T)
```

**Дүрсэлсэн өгөгдлийн тайлбар**
- **Анги** цонх тод **бүсүүдэд** хуваагдана: хамгийн үнэтэй 10% нь **бүхэлдээ** бараан (1-р анги), дараагийн 30%-д 1-р ба 2-р анги холилдож, доод хагас нь бараг бүхэлдээ цайвар (3-р анги). `Fare` ба `Pclass`-ийн хүчтэй хамаарлыг хоёр цонхны пикселийг **ижил байрлалаар** харьцуулж шууд харна. Хамгийн доод мөрөнд цөөн бараан пиксел бий — **£0 үнэтэй** (үнэгүй) 1–2-р ангийн тасалбарууд: эрэмбэлэлт гаж утгыг ингэж илрүүлдэг.
- **Хувь заяа** цонхонд **дээд мөрүүд** (үнэтэй) цэнхэр давамгайтай: хамгийн үнэтэй 10%-ийн **77%** амьд үлдсэн бол хамгийн хямд 10%-ийнх ердөө **13%**. Гэхдээ өөрчлөлт **жигд биш**: дунд хэсэгт (≈ £26–£31) цэнхэр, саарал холилдож, тэр хэсгийн 1-р ангийн эзлэх хувь 29–43% байна.
- **Нас** цонх бараг жигд “шуугиантай” — үнийн дараалалтай тод хамааралгүй. Харин **доод мөрүүдэд** (хямд тасалбар) **саарал** (нас тодорхойгүй) пиксел олон: дутуу насны 177 мөрийн **136** нь 3-р ангийнх. Энэ нь өгөгдлийн **чанарын** асуудлыг (дутуу утга санамсаргүй биш) илрүүлсэн чухал олдвор.

**Аргачлалын давуу тал**
- **Орон зайн хамгийн өндөр хэмнэлттэй** арга: нэг өгөгдлийн утга = нэг пиксел, тиймээс 891 (эсвэл сая сая) мөрийн **бүх хувьсагчийн бүх утгыг** нэгтгэлгүй харуулна.
- Цэг давхцах (over-plotting) асуудал **огт байхгүй**; бүх цонх **ижил эрэмбэ**-тэй тул хувьсагч хоорондын **далд корреляц, бөөгнөрөл, содон утга**, мөн **дутуу утгын бүтэц**-ийг глобал түвшинд алдагдалгүй харна.

*Хязгаарлалт:* утгыг зөвхөн **өнгө**-өөр кодлодог тул нарийн тоон утга уншигдахгүй; эрэмбэлэх хувьсагч болон байрлуулах загварын (мөр мөрөөр, спираль, Hilbert муруй) сонголтоос үр дүн хамаарна.

**Нийт дүгнэлт.** Титаникт амьд үлдэх магадлалыг **хүйс** (эмэгтэй 74%, эрэгтэй 19%), дараа нь **анги ба тасалбарын үнэ** (1-р анги 63%, 3-р анги 24%) тодорхойлсон. Эдгээр хувьсагч хоорондоо **харилцан үйлчилдэг**: 3-р ангийн эмэгтэйчүүд (50%) 1-р ангийн эрэгтэйчүүдээс (37%) илүү амьд үлдсэн. **Нас** ганцаараа сул боловч **хүүхэд байх** нь зөвхөн 1–2-р ангид шийдвэрлэх давуу тал өгсөн (2-р ангийн хөвгүүд 82% vs 3-р ангийн хөвгүүд 23%).

Аргачлал бүр олон хувьсагчтай өгөгдлийн өөр талыг харуулдаг: **ангиллын** хувьсагчдад mosaic ба heatmap, **тоон** хувьсагчдын хамааралд SPLOM ба PCP, **бүлгийн профиль**-д glyph, **бүх өгөгдлийг нэгтгэлгүй** харахад pixel-based, **бүлэг хоорондын харьцуулалт**-д small multiples тохиромжтой. Тиймээс бүрэн дүгнэлтэд хүрэхийн тулд нэг биш, **бие биеэ нөхдөг хэд хэдэн аргачлалыг** хамт хэрэглэх хэрэгтэй.
